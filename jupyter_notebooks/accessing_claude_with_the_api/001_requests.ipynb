## Multi-Turn conversations

```python
message = client.messages.create(
    model = model,
    max_tokens = 1000,
    messages = [
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)

print("First message: \n" + message.content[0].text + "\n")

message = client.messages.create(
    model = model,
    max_tokens = 1000,
    messages = [
        {
            "role": "user",
            "content": "Write another sentence"
        }
    ]
)

print("Second message: \n" + message.content[0].text + "\n")
```

```python
def add_user_message(messages, text):
    user_message = { "role": "user", "content": text }
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = { "role": "assistant", "content": text }
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model = model,
        max_tokens = 1000,
        messages = messages
    )
    return message.content[0].text
```

```python
# Make a starting list of messages
messages = []

# Add the initial user question of "Define quantum computing in one sentence"
add_user_message(messages, "Define quantum computing in one sentence")

# print(messages) -> [{'role': 'user', 'content': 'Define quantum computing in one sentence'}]

# Pass the list of messages into 'chat' to get an answer
answer = chat(messages)

# print(answer) -> **Quantum computing** is a type of computation that harnesses quantum mechanical phenomena — such as superposition and entanglement — to process information in ways that can solve certain complex problems exponentially faster than classical computers.

# Take the answer and add it as an assistant message into our list
add_assistant_message(messages, answer)

# print(messages) -> [
# {'role': 'user', 'content': 'Define quantum computing in one sentence'},
# {'role': 'assistant', 'content': '**Quantum computing** is a form of computation that harnesses quantum mechanical phenomena — such as superposition and entanglement — to process information in ways that can solve certain complex problems exponentially faster than classical computers.'}
# ]

# Add in the user's follow-up question
add_user_message(messages, "Write another sentence")

# Call chat again with the list of messages to get a final answer
answer = chat(messages)

print(answer)
```
